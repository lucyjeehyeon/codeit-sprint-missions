## 문제 1. 3의 배수 출력하기

### 🧩 문제 설명
1부터 n까지의 숫자 중에서 3의 배수만 출력하는 함수를 작성하세요.

### 👉🏻 함수 설명
`print_multiples_of_three(n)`
- `n` (int) : 3의 배수를 구할 범위의 최댓값

### 💡 입출력 예시  

***예시1**

입력: `7`

출력
>```
>3
>6
>```


***예시2**

입력: `30`

출력
>```
>3
>6
>9
>12
>15
>18
>21
>24
>27
>30
>```

````python
def print_multiples_of_three(n):
    for i in range(1, n + 1):  # 1부터 n번째까지 반복
        if i % 3 == 0:  # 위의 범위를 가진 i가 3으로 나눴을 떄 나머지가 0일 때(3의 배수 일때)
           print(i)  # i를 출력한다
````

````python
# 이 코드를 실행해서 함수를 확인합니다
print_multiples_of_three(7)
````

## 문제 3. 모음 제거하기

### 🧩 문제 설명
문자열에서 모음(a, e, i, o, u)을 제거한 문자열을 반환하는 함수를 작성하세요.

### 👉🏻 함수 설명
`remove_vowels(s)`
- `s` (string) : 영어 소문자 알파벳으로 이루어진 문자열

### 💡 입출력 예시

***예시1**
- 입력: `"codeit"`
  
- 출력: `"cdt"`


***예시2**
- 입력: `"python programming"`
  
- 출력: `"pythn prgrmmng"`

````python
def remove_vowels(s):
    list_s = list(s)  # 문자열을 리스트로 변환
    for x in ["a", "e", "i", "o", "u"]:  # 모음 제거 반복문
        while x in list_s:  # 같은 모음이 여러 개 있을 경우 제거가 될 때까지 반복
            list_s.remove(x)
    return "".join(list_s)  # 모음 제거된 문자열 출력
````

````python
# 이 코드를 실행해서 함수를 확인합니다
remove_vowels("codeit")
````

## 문제 4. 숫자 비교하기

### 🧩 문제 설명

두 개의 숫자를 [큰 숫자, 작은 숫자] 순서대로 반환하는 함수를 작성하세요.  
단, 동일한 숫자를 입력한 경우에는 동일한 숫자 2개를 그대로 반환합니다.

### 👉🏻 함수 설명

`compare_numbers(num1, num2)`
- `num1` (int) : 첫 번째 숫자
- `num2` (int) : 두 번째 숫자

### 💡 입출력 예시

***예시1**
- 입력: `20, 30`

- 출력: `30, 20`

***예시2**
- 입력: `10, 10`

- 출력: `10, 10`


````python
def compare_numbers(num1, num2):
    if num1 >= num2:  # num1이 num2보다 크거나 같을 경우
        return f"{num1}, {num2}"  #num1을 먼저 출력
    else:  # 그외의 경우(num1이 num2보다 작을 경우)
        return f"{num2}, {num1}"  #num2 먼저 출력
````

````python
# 이 코드를 실행해서 함수를 확인합니다
print(compare_numbers(20, 30))
print(compare_numbers(10, 10))
````

## 문제 5. 문자열 길이 반환하기

### 🧩 문제 설명

문자열의 길이를 반환하되, 문자열이 비어 있는 경우 `"문자열이 비어 있습니다."` 를 반환하는 함수를 작성하세요.

### 👉🏻 함수 설명

`check_string_length(s)`
- `s` (string) : 영어 알파벳 및 공백이 포함된 문자열

### 💡 입출력 예시

***예시1**

- 입력: `"Python"`
  
- 출력: `6`

***예시2**

- 입력: `""`
  
- 출력: `"문자열이 비어 있습니다."`

````python
def check_string_length(s):
    if s == "":  # s가 ""(빈 문자열)일 경우
        return "문자열이 비어 있습니다." # "문자열이 비어 있습니다." 반환
    else:  # s가 빈 문자열이 아닐 경우
        return len(s)  # 문자열 길이 반환
````

````python
# 이 코드를 실행해서 함수를 확인합니다
print(check_string_length("Python"))
print(check_string_length(""))
````

## 문제 8. 두 숫자 나누기

### 🧩 문제 설명

첫 번째 숫자를 두 번째 숫자로 나누되, 예외가 발생할 경우 `"잘못된 입력입니다."` 를 반환하는 함수를 작성하세요.

### 👉🏻 함수 설명

`divide_numbers(num1, num2)`
- `num1` (int) : 첫 번째 숫자
- `num2` (int) : 두 번째 숫자
- `num1`을 `num2`로 나누는 함수입니다.
- 예외가 발생하면 `"잘못된 입력입니다."`를 반환합니다.
- `try-except` 구문을 활용하는 것도 좋습니다. 잘 모르겠다면 검색을 시도해 보세요.

### 💡 입출력 예시

***예시1**

- 입력: `42, 7`

- 출력: `6`

***예시2**

- 입력: `135, 0`

- 출력: `"잘못된 입력입니다."`


````python
def divide_numbers(num1, num2):
    try:
        return int(num1 / num2)  # num1에 num2를 나눈 정수값을 출력
    except:
        return "잘못된 입력입니다."  # "잘못된 입력입니다." 출력
````

````python
# 이 코드를 실행해서 함수를 확인합니다
print(divide_numbers(42, 7))
print(divide_numbers(135, 0))
````

## 문제 9. BMI 계산하기

### 🧩 문제 설명

사용자가 입력한 키와 몸무게 값을 이용해 BMI를 계산하고 비만도를 판별하는 함수를 작성하세요.

### 👉🏻 함수 설명
`check_obesity(height, weight)`
- `height` (int): 키 정보 (cm)
- `weight` (int): 몸무게 정보 (kg)
- 다음 공식을 사용하여 BMI를 계산합니다.
$$ BMI = \frac{\text{체중 (kg)}}{\text{신장 (m)}^2} $$


- BMI 값을 기준으로 비만도를 다음과 같이 분류하여 반환합니다.
    - 18.5 미만 → '저체중'
    - 18.5 이상 23 미만 → '정상'
    - 23 이상 25 미만 → '과체중'
    - 25 이상 → '비만'

### 💡 입출력 예시

***예시1**

* 입력: `175, 60`
* 출력: `정상`

***예시2**

* 입력: `180, 88`
* 출력: `비만`


````python
def check_obesity(height, weight):
    bmi = weight / (height / 100)**2  # BMI 계산식
    if bmi < 18.5:  # BMI가 18.5 미만인 경우
        return "저체중"  # "저체중" 출력
    elif bmi < 23:  # BMI가 18.5 이상 23 미만인 경우
        return "정상"  # "정상" 출력
    elif bmi < 25:  # BMI가 23 이상 25 미만인 경우
        return "과체중"  # "과체중" 출력
    else:  # 이외의 경우(BMI가 25 이상인 경우)
        return "비만"  # "비만" 출력
````

````python
# 이 코드를 실행해서 함수를 확인합니다
print(check_obesity(175, 60))
print(check_obesity(180, 88))
````

## 문제 10. 문자열 압축하기

### 🧩 문제 설명

주어진 문자열에서 연속해서 같은 문자가 반복되는 경우, 그 문자와 반복 횟수를 이용해 문자열을 압축하는 함수를 작성하세요.
예를 들어 "aabcccccaaa"는 "a2b1c5a3`로 압축될 수 있습니다.
단, 압축 결과의 길이가 원본 문자열보다 짧을 경우에만 압축 결과를 반환하고, 그렇지 않으면 원래 문자열을 반환합니다.

### 👉🏻 함수 설명

`compress_string(s)`
- `s` (string) : 압축하고자 하는 문자열

### 💡 입출력 예시

***예시1**
- 입력: `"aaabbaaa"`
  
- 출력: `"a3b2a3"`

***예시2**
- 입력: `"aabcccccaaa"`
  
- 출력: `"a2b1c5a3"`


````python
# 10번은 못 풀겠어서 해설 참고했습니다!

def compress_string(s):
    count = 1
    compressed_string = ""

    for i in range(len(s)):
        if (i+1 < len(s)) and (s[i] == s[i+1]):
            count += 1
        else:
            compressed_string += s[i] + str(count)
            count = 1

    if len(compressed_string) < len(s):
        return compressed_string
    else:
        return s
````

````python
# 이 코드를 실행해서 함수를 확인합니다
compress_string("aaabbaaa")
````

````python
from random import randint

def generate_numbers():
    x = randint(0, 9)
    y = randint(0, 9)
    if x != y:
        z = randint(0, 9)
        if z != x and z != y == True:
            solution = [x, y, z]
            return solution
        else:
            while z == x or z == y == True:
                z = randint(0, 9)
            solution = [x, y, z]
            return solution
    else:
        while x == y:
            y = randint(0, 9)
        z = randint(0, 9)
        if z != x and z != y == True:
            solution = [x, y, z]
            return solution
        else:
            while z == x or z == y == True:
                z = randint(0, 9)
            solution = [x, y, z]
            return solution



def take_guess():
    print("숫자 3개를 하나씩 차례대로 입력하세요.")
    a = int(input("1번째 숫자를 입력하세요.:"))
    if a >= 0 and a <= 9:
        b = int(input("2번째 숫자를 입력하세요.:"))
        if b >= 0 and b <= 9 and a != b:
            c = int(input("3번째 숫자를 입력하세요.:"))
            if c >= 0 and c <= 9 and a != c and b != c:
                guess = [a, b, c]
                return guess
            else:
                while a == c or b == c or (not(c >= 0 and c <=9)):
                    if a == c or b == c:
                        print("중복되는 숫자 입니다. 다시 입력하세요.")
                    else:
                        print("범위를 벗어나는 숫자입니다. 다시 입력하세요.")
                    c = int(input("3번째 숫자를 입력하세요.:"))
                guess = [a, b, c]
                return guess
        else:
            while a == b or (not(b >= 0 and b <=9)):
                if a == b:
                    print("중복되는 숫자 입니다. 다시 입력하세요.")
                else:
                    print("범위를 벗어나는 숫자입니다. 다시 입력하세요.")
                b = int(input("2번째 숫자를 입력하세요.:"))
            c = int(input("3번째 숫자를 입력하세요.:"))
            if c >= 0 and c <= 9 and a != c and b != c:
                guess = [a, b, c]
                return guess
            else:
                while a == c or b == c or (not(c >= 0 and c <= 9)):
                    if a == c or b == c:
                        print("중복되는 숫자 입니다. 다시 입력하세요.")
                    else:
                        print("범위를 벗어나는 숫자입니다. 다시 입력하세요.")
                    c = int(input("3번째 숫자를 입력하세요.:"))
            guess = [a, b, c]
            return guess
    else:
        while not(a >= 0 and a <= 9):
            print("범위를 벗어나는 숫자입니다. 다시 입력하세요.")
            a = int(input("1번째 숫자를 입력하세요.:"))
        b = int(input("2번째 숫자를 입력하세요.:"))
        if b >= 0 and b <= 9 and a != b:
            c = int(input("3번째 숫자를 입력하세요.:"))
            if c >= 0 and c <= 9 and a != c and b != c:
                guess = [a, b, c]
                return guess
            else:
                while a == c or b == c or (not(c >= 0 and c <= 9)):
                    if a == c or b == c:
                        print("중복되는 숫자 입니다. 다시 입력하세요.")
                    else:
                        print("범위를 벗어나는 숫자입니다. 다시 입력하세요.")
                    c = int(input("3번째 숫자를 입력하세요.:"))
            guess = [a, b, c]
            return guess
        else:
            while a == b or not(b >= 0 and b <= 9):
                if a == b:
                    print("중복되는 숫자 입니다. 다시 입력하세요.")
                else:
                    print("범위를 벗어나는 숫자입니다. 다시 입력하세요.")
                b = int(input("2번째 숫자를 입력하세요.:"))
            c = int(input("3번째 숫자를 입력하세요.:"))
            if c >= 0 and c <= 9 and a != c and b != c:
                guess = [a, b, c]
                return guess
            else:
                while a == c or b == c or (not(c >= 0 and c <= 9)):
                    if a == c or b == c:
                        print("중복되는 숫자 입니다. 다시 입력하세요.")
                    else:
                        print("범위를 벗어나는 숫자입니다. 다시 입력하세요.")
                    c = int(input("3번째 숫자를 입력하세요.:"))
                guess = [a, b, c]
                return guess


def get_score(guess, solution):
    strike_count = 0
    ball_count = 0
    for i in range(3):
        if guess[i] == solution[i]:
            strike_count += 1
    if guess[0] ==  solution[1]:
        ball_count +=1
    if guess[0] ==  solution[2]:
        ball_count +=1
    if guess[1] ==  solution[0]:
        ball_count +=1
    if guess[1] ==  solution[2]:
        ball_count +=1
    if guess[2] ==  solution[0]:
        ball_count +=1
    if guess[2] ==  solution[1]:
        ball_count +=1

    return strike_count, ball_count
````

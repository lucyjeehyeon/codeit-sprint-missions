### 문제 1-4. 쇼핑몰 고객 데이터 분석

한 쇼핑몰에서 고객의 구매 정보를 2차원 NumPy Array로 관리하고 있다고 합니다. 행은 각각의 고객을 나타내며, 열은 각 주의 구매 금액(만원 단위)을 나타냅니다. (예: 1행 1열은 첫 번째 고객의 첫 주 구매 금액)  
이 쇼핑몰에서 특정 조건을 만족하는 고객들에게 특별 할인 쿠폰을 제공하려고 합니다. 다음 조건 중 하나 이상을 만족하는 고객들을 선별해, 그들의 구매 정보만 담은 새로운 NumPy Array를 출력하세요.  
이때, 새로운 Array에서는 100 이상의 값은 1로, 그보다 작은 값은 0으로 변경해 출력해야 합니다. 고객의 순서는 원본 Array와 달라져도 괜찮아요.

- 특정 주간에 구매 금액이 100 이상인 고객
- 모든 주간의 총 구매 금액이 300 이상인 고객

---
*구매 금액이 100 이상인 주간이 하나라도 있었는지 검사할 때에는 `np.any()`라는 함수를 이용할 수도 있습니다.   
물론 그 외에 Boolean Indexing이나 최댓값 검사 등 다른 방법도 가능하니 알고 있는 방법 내에서 풀어보는 것도 좋아요.

```python
# 구매 데이터
purchase_data = np.array([
    [50, 60, 90, 80, 20],
    [120, 150, 200, 90, 100],
    [30, 20, 10, 40, 50],
    [20, 30, 120, 10, 50],
    [10, 30, 90, 70, 40],
    [90, 80, 99, 70, 60],
    [110, 20, 90, 30, 10]
])
```

```python
# 특정 주간에 구매 금액이 100 이상인 고객
condition = np.any(purchase_data >= 100, axis=1)
condition1 = np.where(condition == True)
condition1
```

```python
# 모든 주간의 총 구매 금액이 300 이상인 고객
condition2 = np.where(purchase_data.sum(axis=1) >= 300)
condition2
```

```python
# 조건 중 하나 이상을 만족하는 고객들을 선별
coupon = []
for c in [condition1, condition2]:
    for i in range(c[0].size):
        if int(c[0][i]) not in coupon:
            coupon.append(int(c[0][i]))

coupon.sort()
coupon
```

```python
# 새로운 Array에서는 100 이상의 값은 1로, 그보다 작은 값은 0으로 변경해 출력
a = []
for i in coupon:
    a.append(np.where(purchase_data[i] >= 100, 1, 0))

adjusted_purchase_data = np.array(a)
```

```python
# 조건에 따라 조정된 Array를 출력하세요
adjusted_purchase_data
```

### 문제 2-7. 타입 변환
retail_orders 데이터셋의 `amount` 컬럼은 숫자 타입이어야 하지만, '3,990'처럼 중간에 쉼표가 포함되거나 '$890'처럼 달러 표시가 포함된 값이 있어 문자열처럼 인식되고 있습니다.  
아래 절차에 따라 컬럼의 타입을 변환해 주세요. 변환 전후의 데이터 타입을 출력해서 비교하면 처리가 잘 되었다는 걸 확인할 수 있습니다.

1. `str.replace()`로 쉼표(,)와 달러 표시($)를 제거해 주세요.
2. 타입을 'float'로 바꿔주세요.
3. 결측값을 중앙값으로 채우세요.
4. 이제 다시 타입을 'int'로 바꿔주세요.

```python
# 먼저 변환 이전의 데이터 타입을 출력합니다
retail_df['amount'].dtype
```

NumPy를 활용해 가상의 고객 데이터를 직접 만들어봅니다.

1. **고객 번호 생성**: 고객 번호를 1부터 100까지 생성하세요. (`np.arange()` 활용)
2. **고객 나이 생성**: 고객의 나이를 18세에서 64세 사이의 정수로 무작위 생성하세요. (`np.random.randint()` 활용)
3. **거주 지역 생성**: 고객이 거주하는 지역을 "서울", "부산", "대구", "인천" 중에서 무작위로 할당하세요. (`np.random.choice()` 활용)
4. **주문 금액 생성**: 고객의 주문 금액을 평균 20만원, 표준편차 8만원인 정규분포로 생성하세요. (`np.random.normal(평균, 표준편차, 개수)` 활용)
   - 이때, `.round(-2)`로 반올림하면 더욱 실제 금액처럼 보이는 값을 얻을 수 있습니다.
6. **가입일 생성**: 2024년 1월 1일을 기준으로 180일 이내의 임의의 날짜 간격을 더해 생성하세요.    
   - `pd.to_datetime()`으로 시작 날짜를 만든 후 `pd.to_timedelta()`에 `np.random.randint()`를 더해 임의의 날짜 간격을 더하면 됩니다.
7. 생성한 데이터를 하나의 DataFrame으로 합치세요.

```python
import numpy as np
import pandas as pd

# 랜덤 시드 설정: 매번 같은 값이 나오도록 고정
np.random.seed(42)

#  고객 수 설정
n = 100

# 각 컬럼 생성
customer_id = np.arange(1,101)
age =  np.random.randint(18, 65, 100)
region =  np.random.choice(['서울', '부산', '대구', '인천'], 100)
order_amount =  np.random.normal(200000, 80000, 100).round(-2)
signup_date =  pd.to_datetime(pd.to_datetime('2024-01-01') + pd.to_timedelta(np.random.randint(0, 181, 100),unit='D'))

# 생성한 데이터를 하나의 DataFrame으로 합치는 코드를 입력하세요
df = pd.DataFrame({'customer_id':customer_id,
                  'age':age,
                  'region':region,
                  'order_amount':order_amount,
                  'signup_date':signup_date})

# 상위 5개 로우 출력하기
df.head()
```

2. 주문 금액(`order_amount`) 컬럼에서 극단적인 값을 제거해 보세요. Q1 - 1.5 * IQR 미만 혹은 Q3 + 1.5 * IQR 초과 범위의 데이터를 이상치로 판단하고 제거합니다.
   (사분위수 계산을 위해서는 `df.quantile()`을 쓸 수 있습니다)

```python
# 데이터의 구조 및 주요 통계량을 확인하는 코드를 입력하세요
df.shape
```

```python
# 주문 금액에서 이상치를 제거하는 코드를 입력하세요
q1 = df['order_amount'].quantile(0.25)
q3 = df['order_amount'].quantile(0.75)

iqr = q3 - q1
lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

outlier = (df['order_amount'] < lower_limit) | (df['order_amount'] > upper_limit)
df = df.query('~ @outlier')
```

```python
# 이상치 제거 후 주문 금액 컬럼의 값 분포를 다시 확인해 봅니다 (describe() 사용)
df.describe()
```

```python
# 지역별 연령대별 평균 주문 금액을 집계하는 코드를 입력하세요
df_group1 = df.groupby(['age_group', 'region'])['order_amount'].mean().reset_index()
```
